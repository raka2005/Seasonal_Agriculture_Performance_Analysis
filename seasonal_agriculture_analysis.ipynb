```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv("seasonal_agriculture_performance_dataset.csv")
data.head()
```

```python
print("Shape of dataset:", data.shape)
print("\nColumn names:")
print(data.columns)
print("\nDataset information:")
data.info()
```

```python
print("\nColumn Names:")
print(data.columns.tolist())
print("\nData Types:")
print(data.dtypes)
print("\nStatistical Summary:")
display(data.describe())
print("\nSeasons:")
print(data["Season"].unique())
print("\nCrops:")
print(data["Crop"].unique())
```

```python
print("Missing values : ")
print(data.isnull().sum())
print("\nDuplicate Rows:")
print(data.duplicated().sum())
print("\nInfinite Values:")
print(data.select_dtypes(include="number").isin([float("inf"), float("-inf")]).sum())
```

```python
season_performance=data.groupby("Season")[[
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Profit_INR"
]].mean()
season_performance
display(season_performance)
```

```python
season_performance.plot(kind="bar",figsize=(5,5))
plt.title("Agricultural performance across seasons")
plt.xlabel("Season")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.show()
```

```python
season_trends = data.groupby("Season")[[
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Fertilizer_kg_ha",
    "Water_Used_m3",
    "Disease_Pest_Risk_pct"
]].mean()
display(season_trends)
```

```python
season_trends.plot(kind="bar",figsize=(5,5))
plt.title("Seasonal patterns in agricultural conditions")
plt.xlabel("Season")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.show()
```

```python
correlation=data[[
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Fertilizer_kg_ha",
    "Water_Used_m3",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Profit_INR"
]].corr()
display(correlation["Yield_Tonnes_Ha"].sort_values(ascending=False))
```

```python
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".3f")
plt.title("Relationship Between Environmental Conditions and Agricultural Performance")
plt.show()
```

```python
crop_season=data.groupby(["Season","Crop"])["Yield_Tonnes_Ha"].mean().unstack()
display(crop_season)
```

```python
crop_season.plot(kind="bar",figsize=(5,5))
plt.title("Crop Yield Comparison Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Yield")
plt.xticks(rotation=0)
plt.legend(title="Crop")
plt.show()
```

```python
from scipy.stats import f_oneway
groups = [
    data[data["Season"] == season]["Yield_Tonnes_Ha"].dropna()
    for season in data["Season"].unique()
]
f_stat, p_value = f_oneway(*groups)
print("ANOVA F-statistic:", f_stat)
print("ANOVA p-value:", p_value)
sns.boxplot(x="Season", y="Yield_Tonnes_Ha", data=data)
plt.title("Yield Distribution and Unusual Values Across Seasons")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.show()
```

### Interpretation of Findings

1. Kharif season shows the highest average yield, production, and profit.
2. Rabi season shows moderate agricultural performance compared with Kharif.
3. Zaid season has the lowest average yield and production and shows negative average profit.
4. Environmental conditions vary across seasons and show relationships with agricultural performance.
5. Crop performance differs across seasons.
6. Yield values show variations across different seasons.

### Conclusion
1. Kharif season shows the best overall agricultural performance.
2. Rabi season shows moderate performance.
3. Zaid season shows the weakest performance, with the lowest average yield and production and negative average profit.
4. Crop performance varies across different seasons.
5. Environmental conditions and resource usage vary between seasons and can influence agricultural outcomes.
